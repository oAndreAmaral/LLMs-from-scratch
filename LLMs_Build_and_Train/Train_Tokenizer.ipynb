```python
# Tokeinizer - Convert the text to tokens and then the tokens to numbers
# Each token has an associated numerical ID so both tokens and IDs are the same thing

import sentencepiece as spm
import os

vocab_size = 4096 # Large vocabulary leads to more precision and complex interligations and more computational complex
```

```python
### Byte Pair Encoding (BPE) ###
# merge the most frequent pairs of bytes to form subwords until the desired vocabulary size is reached. 
# it can split rare words into more frequent subwords units improving the model ability to generalize.

spm.SentencePieceTrainer.train(
    input='wiki.txt',
    model_prefix="wiki_tokenizer_test",
    model_type="bpe",
    vocab_size=vocab_size,
    self_test_sample_size=0,
    input_format="text",
    character_coverage= 0.995, #propotion of the chars in the corpus considered when we build the tokeinzer model, useful for chinese and japones chars
    num_threads= os.cpu_count(),
    split_digits=True,
    allow_whitespace_only_pieces=True,
    byte_fallback=True,
    unk_surface=r"\342\201\207 ",
    normalization_rule_name='identity'
)

print("Tokenizer training completed")
```
